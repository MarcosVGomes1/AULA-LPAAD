## Problema Final – Processamento de Encomendas


```python
def calcular_frete(cliente, *pesos, **informacoes):
    # Desenvolva a solucao aqui

    if not isinstance (cliente, str) or not cliente:
        print("Nome inválido")
        return
    if not pesos:
        print("Peso inválido, peso não informado")
        return
    for peso in pesos:
        if not isinstance(peso, (int, float)):
            print("Peso inválido, nao é int ou float")
            return
        if peso < 1 or peso >= 30:
            print("Peso inválido, menor que 1 ou maior que 30")
            return
        if peso * len(pesos) > 100:
            print("Peso inválido, total maior que 100")
            return
    if "regiao" in informacoes:
        regioes = ["Sudeste", "Sul", "Centro-Oeste", "Nordeste", "Norte"]
        if informacoes["regiao"] not in regioes:
            print("Regiao inválida")
            return
    if "modalidade" in informacoes:
        # economica sem acréscimo * 1 - Expressa acréscimo de 40% * 1.4
        modalidades = ["economica", "expressa"]
        if not isinstance(informacoes["modalidade"], (str)) or not ["modalidade"]:
            print("Modalidade inválida, string or not")
            return
        if informacoes["modalidade"] not in modalidades:
            print("Modalidade inválida, sem modalidade")
            return
    if "seguro" in informacoes:
        # com seguro + 5% * 0.05
        if not isinstance(informacoes["seguro"], bool):
            return "Informação do seguro inválida"
        
    # calculo do frete

    base_regiao ={"Sudeste" : 5.00,
                  "Sul": 6.00,
                  "Centro-oeste": 7.00,
                  "Nordeste": 8.00,
                  "Norte": 10.00}
    
    tarifa_frete = base_regiao[informacoes["regiao"]]
    total_peso = sum(pesos)

    frete_base = total_peso * tarifa_frete

    if "modalidade" in informacoes:
        if "modalidade" == 'economica':
            total_frete = frete_base * 1.00
        else:
            total_frete = frete_base * 1.4

    if "seguro" in informacoes:
        if "seguro" == True:
            total_completo = total_frete * 1.05
        else:
            total_completo = total_frete * 1.00
        valor_seguro = total_frete * 0.05

    print(f"Resultados finais:")
    print(f"nome: {cliente}")
    print(f"Peso Total: {total_peso} kg")
    print(f"Frete-Base: R$ {frete_base}")
    print(f"Frete com modalidade: R$ {total_frete:.2f}")
    print(f"Seguro: R$ {valor_seguro:.2f}")
    print(f"Frete Final: R$ {total_completo:.2f}")
```

```python
calcular_frete("Carlos", 5, 10, 3, regiao="Sudeste",modalidade="expressa",seguro=True)
```
