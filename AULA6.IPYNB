*amplitude* - diferença entre o maior e menor

*variância* -mede o quanto os valores se afastam da média

```python
def calcula_variancia (lista):
    media = calcula_media(lista)
    soma_diferenca = 0

    
    for i in lista:
        diferenca = i - media
        diferenca_ao_quadrado = diferenca**2
        soma_diferenca += diferenca_ao_quadrado

    variancia = soma_diferenca/len(lista)
    return variancia
```

Amplitude

```python
def calcula_amplitude(lista):
    maximo = max(lista)
    minimo = min(lista)

    amplitude = maximo - minimo

    return amplitude
```

```python
vendas = [18500,22100,19800,24300,25600,21700]

calcula_amplitude(vendas)
```

Quartis

```python
def calcula_quartis(lista):
    lista_ordenada = sorted(lista)
    tamanho = len(lista)

    posicao_q1 = 1 * (tamanho + 1) // 4 
    posicao_q2 = 2 * (tamanho + 1) // 4 
    posicao_q3 = 3 * (tamanho + 1) // 4 

    q1 = lista_ordenada[int(posicao_q1) -1]
    q2 = lista_ordenada[int(posicao_q2) -1]
    q3 = lista_ordenada[int(posicao_q3) -1]

    return q1, q2, q3

vendas = [42,18,35,27,50,31,22]

calcula_quartis(vendas)
```
